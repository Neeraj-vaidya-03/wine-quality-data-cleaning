```python
df_cleaned = df.dropna()
print("Row before and after dropping nulls:", len(df), len(df_cleaned))
```

```python
df_filled = df.fillna(0)
```

```python
df_mean_filled = df.fillna(df.mean(numeric_only=True))
df_mean_filled
```

```python
df.columns = df.columns.str.strip()
df.columns
```

```python
df.columns =df.columns.str.replace(' ', '_')
df.columns
```

```python
df.columns = df.columns.str.lower()
df.columns
```

```python
df = df.rename(columns={'quality': 'wine_quality'})
df 
```

```python
print("Total duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print("New shape after removing duplicates:", df.shape)
```

```python
high_quality = df[df['wine_quality'] > 7]
print(f"Number of high-quality wines: {len(high_quality)}")
```

```python
for index, col in enumerate(df.columns):
    print(f"Column {index}: '{col}'")
```

```python
df['quality'] = df['wine_quality'].apply(lambda x: 'High' if x > 7 else 'Low')
df['quality'].value_counts()
```

```python
df['quality'] = df['quality'].astype(str)
df['quality'] = df['quality'].map({'High': 1, 'Low': 0})
df['quality']
```

```python
df['quality'] = df['quality'].astype(str)
df['quality']  
```

```python
df['quality'] = df['quality'].replace('6', 'Six')
df['quality']
```

```python
df = df.rename(columns={'fixed_acidity': 'fa', 'volatile_acidity': 'va'})
df
```

```python
df= df.drop(columns=['fa'])
df 
```

```python
print(df.memory_usage(deep=True))
```

```python
import numpy as np
df['alcohol_category'] = np.where(df['alcohol'] > 10.0, 'High', 'Normal')
df['alcohol_category']
```

```python

bins = [0, 9.0, 11.0, 15.0]
labels = ['Low', 'Medium', 'High']
df['alcohol_bin'] = pd.cut(df['alcohol'], bins=bins, labels=labels)
df['alcohol_bin']
```

```python
sample_df = df.sample(n=6)
display(sample_df)
```

```python
df['high_sugar_flag'] = df ['residual_sugar'] > 10.0 
df['high_sugar_flag']
```

```python
print(df.corr(numeric_only=True))
```

```python
Q1 = df['residual_sugar'].quantile(0.25)
Q3 = df['residual_sugar'].quantile(0.75)
IQR = Q3 - Q1
print(f"IQR for residual sugar is: {IQR}")
```

```python
import matplotlib.pyplot as plt 

plt.figure(figsize=(10,6))
scatter = plt.scatter(
    df['alcohol'],
    df['quality'],
    c=df['residual_sugar'],
    cmap='viridis',
    alpha=0.6
)



plt.colorbar(scatter, label='Residual sugar')
plt.xlabel('Alcohol')
plt.ylabel('Quality')
plt.title('Alcohol VS Quality (colormap: viridis)')
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10,6))
corr_matrix = df.corr(numeric_only=True)

plt.imshow(corr_matrix, cmap='coolwarm', aspect='auto', vmin= -1, vmax=1)
plt.colorbar(label='correlation coefficient')
plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=45)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)
plt.title('Wine Dataset Correlation Heatmap (Colormap: coolwarm)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df_cleaned['fixed acidity'], bins=30, color='teal', edgecolor='black', alpha=0.8)
plt.xlabel('Fixed Acidity')
plt.ylabel('Frequency')
plt.title('Distribution of Fixed Acidity')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df_cleaned['volatile acidity'], vert=False, patch_artist=True, 
            boxprops=dict(facecolor='orange', color='black'),
            medianprops=dict(color='black', linewidth=2))
plt.xlabel('Volatile Acidity')
plt.title('Box Plot of Volatile Acidity (Checking Outliers)')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```
